```python
def numw(tsif):

    s =['ноль','один', 'два','три','четыре ','пять', 'шесть', 'семь', 'восемь','девять','десять','одиннадцать','двенадцать','тринадцать','четрынадцать', 
    'пятнадцать','шестнадцать','семнадцать','восемнадцать','девятнадцать','двадцать', '']
    s2 =['двадцать', 'тридцать', 'сорок', 'пятьдесят', 'шестьдесят', 'семьдесят', 'восемьдесят', 'девяносто']
    if tsif <= 20:
        return  s[tsif ]
    else:
        return s2[tsif // 10 - 2] + ' ' + s [tsif %10 ]
print('Введите число от 0 до 99:')
user_input = input()
if user_input.isdigit():
    n = int(user_input)
    if 0<=n<=99:
        print(numw(n))
    else:
        print('ошибка, число должно быть от 0 до 99')
else:
    print('ошибка, введите ЧИСЛО')
```
